### RANSAC

```python
# helper functions
def apply_mask(PC, mask):
    """Applies a defined mask to the point cloud."""
    rows, cols = np.where(mask)
    PC_masked = PC[rows, cols, :].astype(float)
    return PC_masked, rows, cols

def fit_plane(points):
    """ 
    Takes 3 points (3x3 array) and fits a plane. Returns 
    unit normal n and offset d.
    """
    p1, p2, p3 = points
    # form edge vectors
    vec1 = p2 - p1
    vec2 = p3 - p1

    n_raw = np.cross(vec1, vec2)
    normal_len = np.linalg.norm(n_raw)

    # if collinear points, return None
    if normal_len == 0:
        return None, None, False
    
    n = n_raw / normal_len
    d = -float(np.dot(n, p1))
    
    return n, d, True

def calculate_distance(points, n, d):
    """Finds the point to plane distances for multiple points (N,3) at once."""
    values = np.abs(np.dot(points, n) + d)
    return values

def find_inliers(distances, threshold):
    """Finds inliers based on a distance threshold."""
    inliers_mask = distances < threshold
    inliers_idx = np.where(inliers_mask)[0]
    inliers_count = inliers_mask.sum()

    return inliers_mask, inliers_idx, int(inliers_count)

def sample_points(p):
    """ Returns three distinct integers (indices) from 0..p-1"""
    return np.random.choice(p, 3, replace=False)
```

```python
def RANSAC(PC, max_iter, distance_thresh):
    best_count = -1     # becuase number of inliers can be zero
    best_n = None
    best_d = None
    best_inliers = None

    if PC.shape[0] < 3:
        raise ValueError("Not enough points to fit a plane.")
    
    for iter in range(max_iter):
        points = sample_points(PC.shape[0])
        n, d, valid = fit_plane(PC[points, :])

        # skip collinear points
        if not valid:
            continue

        distances = calculate_distance(PC, n, d)
        inliers, idx, inlier_count = find_inliers(distances, distance_thresh)

        if inlier_count > best_count:
            best_count = inlier_count
            best_n = n
            best_d = d
            best_inliers = inliers
        
    return best_n, best_d, best_inliers, best_count
    
```

### Discussion

1. Our current implement assumes that after removing the floor, the second most largest connected region is the box, so this implementation might fail if we have more larger connected region. A potential solution then might be to filter the dataset even further by adding a max and min distance threshold.

2. The performance of RANSAC is also heavily dependant on the choice of our distance threshold and since we are randomly sampling, choosing a good threshold was tricky. This can be dealt with better by adding an adaptive threshold which adjusts the threshold if number of inliers is too less as well as adding MLESAC for inlier scoring.

```python
def dist_based_inliers(distances, threshold, gamma):
    """Finds inliers based on a distance threshold."""
    inliers_mask = distances < threshold
    inlier_sum = np.sum(distances[inliers_mask])
    outlier_mask = distances >= threshold
    outlier_sum = outlier_mask.sum()
    total_cost = inlier_sum + (outlier_sum * gamma)
    return inliers_mask, total_cost
```

```python
def MLESAC(PC, max_iter, distance_thresh, gamma):
    best_cost = math.inf     # we have to minimize the cost hence we took infinity
    best_n = None
    best_d = None
    best_inliers_mask = None
    best_inlier_count = 0

    if PC.shape[0] < 3:
        raise ValueError("Not enough points to fit a plane.")
    
    for iter in range(max_iter):
        points = sample_points(PC.shape[0])
        n, d, valid = fit_plane(PC[points, :])

        # skip collinear points
        if not valid:
            continue

        distances = calculate_distance(PC, n, d)
        inliers_mask, total_cost = dist_based_inliers(distances, distance_thresh, gamma)

        if total_cost < best_cost:
            best_cost = total_cost
            best_n = n
            best_d = d
            best_inliers_mask = inliers_mask
            best_inlier_count = inliers_mask.sum()
        
    return best_n, best_d, best_inliers_mask, best_cost, best_inlier_count
```

```python
from scipy.ndimage import binary_opening, binary_closing, label
import matplotlib.pyplot as plt

def plot_box_mlesac(PC, floor_inlier_mask, rows, cols, D_vis, valid_mask):
    """
    1. Cleans the floor mask.
    2. Removes floor points to find candidates.
    3. Fits a plane to the box top.
    4. Extracts the largest connected component for the box.
    """
    H, W = PC.shape[:2]

    # Create and Clean Floor Mask
    floor_mask = np.zeros((H, W), dtype=bool)
    # Map the 1D inliers back to the 2D image
    floor_mask[rows[floor_inlier_mask], cols[floor_inlier_mask]] = True

    # Morphological cleaning
    kernel_open = np.ones((5,5), dtype=bool)
    kernel_close = np.ones((3,3), dtype=bool)

    floor_mask_clean = binary_opening(floor_mask, structure=kernel_open)
    floor_mask_clean = binary_closing(floor_mask_clean, structure=kernel_close)


    # Isolates Non-Floor Points
    # valid_mask is the original mask of valid depth pixels
    keep_nonfloor = valid_mask & ~floor_mask_clean
    
    # Extracts these specific points for the second RANSAC pass
    PC_box_candidates, r_box, c_box = apply_mask(PC, keep_nonfloor)
    
    # Find Box Top Plane
    # Using MLESAC 
    print("Fitting box top...")
    n_top, d_top, top_in_mask, top_cost, top_count = MLESAC(
        PC_box_candidates, 
        max_iter=2000, 
        distance_thresh=0.007, # Tighter threshold for the box 
        gamma=0.5
    )

    # Create Box Mask & Connected Components
    top_mask = np.zeros((H, W), dtype=bool)
    top_mask[r_box[top_in_mask], c_box[top_in_mask]] = True

    # Find largest connected component
    l, num_connected = label(top_mask)
    l_sizes = np.bincount(l.ravel())
    l_sizes[0] = 0 # Ignore background
    box_top_mask = (l == l_sizes.argmax())


    # Final cleanup of the box mask
    box_top_clean = binary_opening(box_top_mask, np.ones((3,3), bool))
    box_top_clean = binary_closing(box_top_clean, np.ones((3,3), bool))

    # Visualization
    plt.figure(figsize=(6,6))
    plt.imshow(D_vis, cmap='gray')
    plt.imshow(box_top_clean, cmap='spring', alpha=0.4)
    plt.title("Box Top")
    plt.axis('off')

    plt.tight_layout()
    plt.show()

    return box_top_clean, n_top, d_top

```

Discussion: From the results I found that MLESAC is less susceptible to hyperparameter choices. Because standard RANSAC performs binary classification that is a point is inlier or outlier and because of this we might miss a better plane just because the other plane has more number of inliers. In MLESAC alorithm we assign cost to to the points based on the distance from the plane

Advantages:
1. High Accuracy: By Minimizing the distance errors of inliers, it finds a plane that better represents the box top or floor.
2. Stability: From the results we can see that the inlier count remains stable even when the cost parameters vary.

Disadvantages:
1. New Hyperparameter: MLESAC introduces $\gamma$ (outlier penalty), which adds another variable to tune. If $\gamma < \epsilon$, the logic breaks.
2. Complexity: In this we need to calculate the distance for every inlier to sum them up, whereas standard RANSAC only requires a boolean check (distance < threshold).

```python
def preemptive_ransac(PC, M, B, threshold, gamma):
    normal_vectors = []
    planes = []
    for _ in range(M):
        points = sample_points(PC.shape[0])
        n, d, valid = fit_plane(PC[points, :])
        while not valid:
            points = sample_points(PC.shape[0])
            n, d, valid = fit_plane(PC[points, :])
        normal_vectors.append(n)
        planes.append(d)
    PC_shufled = PC.copy()
    np.random.shuffle(PC_shufled)  # Randomly shuffles the point cloud for better results
    active_indices = list(range(M))
    hypotheses_costs = np.zeros(M)
    for p in range(PC_shufled.shape[0]):
        for h in active_indices:
            dist = calculate_distance(PC_shufled[p], normal_vectors[h], planes[h])
            point_cost = dist if dist < threshold else gamma
            hypotheses_costs[h] +=point_cost
        i = p + 1
        if (i % B ==0):
            function = int(M * 2**(-math.floor(i / B)))
            no_of_hypos_keep = max(1, function)
            active_indices.sort(key= lambda idx: hypotheses_costs[idx])
            active_indices = active_indices[:no_of_hypos_keep]
        if len(active_indices) == 1: # If only one plane is left we select that plane
            break
    best_plane_idx = active_indices[0]
    best_n = normal_vectors[best_plane_idx]
    best_d = planes[best_plane_idx]
    best_cost = hypotheses_costs[best_plane_idx]
    all_distances = calculate_distance(PC, best_n, best_d)
    final_box_top_mask = all_distances< threshold
    return best_n, best_d, best_cost, final_box_top_mask
```

Discussion: From the results I found that increasing the hypothesis budget M from 128 to 1024 improved the accuracy of the estimated plane, as a larger initial pool increases the probability of capturing the optimal plane. Which we can also see from the above plotting of the box top.

Advantages:
1. Real-Time Predictability: The standard RANSAC relies on the max_iter which is provided by the user, whereas in this one we know exactly how many iterations will occur.
2. Efficiency: It stops wasting resources on bad models immediately. A model that doesn't fit the first 100 points is discarded, whereas standard RANSAC might waste time evaluating it against all points.

Disadvantages

1. Greedy Pruning: The main disadvantage is that it might discarding the best plane early. Because if the best model contains less inliers from the first batch of the points.
2. Memory Usage: It need to store all the M planes and there running costs, Whereas in standard RANSAC we only need to store the current best model.
